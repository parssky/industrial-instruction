```python
def remove_complex_answers(samples):
    """
    Remove samples with complex answers (like model numbers) from the list.
    Returns a new list with only samples containing simple answers.
    """
    filtered_samples = []
    removed_samples = []
    
    for i, sample in enumerate(samples):
        answer = sample.get('a*', [])
        
        # Check if answer is valid and simple
        is_simple = True
        
        if not isinstance(answer, list):
            is_simple = False
        else:
            for item in answer:
                item_str = str(item)
                
                # Complex patterns to filter out:
                # 1. Contains hyphens (model numbers)
                # 2. Too long (> 3 characters)
                # 3. Looks like a product code
                if ('-' in item_str or 
                    len(item_str) > 3 or
                    (any(c.isdigit() for c in item_str) and 
                     any(c.isalpha() for c in item_str) and 
                     len(item_str) > 2)):
                    is_simple = False
                    break
        
        if is_simple:
            filtered_samples.append(sample)
        else:
            removed_samples.append((i, sample))
    
    print(f"Original samples: {len(samples)}")
    print(f"Filtered samples: {len(filtered_samples)}")
    print(f"Removed samples: {len(removed_samples)}")
    
    return filtered_samples, removed_samples
```

```python
# Usage
r0_samples, removed = remove_complex_answers(r0_samples)
print("\nRemoved samples examples:")
for idx, sample in removed[:3]:
    print(f"  Sample {idx}: {sample.get('a*')}")

r1_samples, removed = remove_complex_answers(r1_samples)
print("\nRemoved samples examples:")
for idx, sample in removed[:3]:
    print(f"  Sample {idx}: {sample.get('a*')}")

r2_samples, removed = remove_complex_answers(r2_samples)
print("\nRemoved samples examples:")
for idx, sample in removed[:3]:
    print(f"  Sample {idx}: {sample.get('a*')}")

r3_samples, removed = remove_complex_answers(r3_samples)
print("\nRemoved samples examples:")
for idx, sample in removed[:3]:
    print(f"  Sample {idx}: {sample.get('a*')}")

r4_samples, removed = remove_complex_answers(r4_samples)
print("\nRemoved samples examples:")
for idx, sample in removed[:3]:
    print(f"  Sample {idx}: {sample.get('a*')}")
```

```python
from datasets import load_from_disk, concatenate_datasets, DatasetDict, load_dataset

def normalize(example):
    # answer
    if example["answer"] is None:
        example["answer"] = []
    elif isinstance(example["answer"], str):
        example["answer"] = [example["answer"]]

    # documents
    if example["documents"] is None:
        example["documents"] = []
    elif isinstance(example["documents"], str):
        example["documents"] = [example["documents"]]

    return example

for split in ['r0', 'r1', 'r2', 'r3', 'r4']:
    dataset_dict[split] = dataset_dict[split].map(normalize)


train_datasets = []
test_datasets = []

for split in ['r0', 'r1', 'r2', 'r3', 'r4']:
    # Shuffle the dataset
    shuffled = dataset_dict[split].shuffle(seed=42)
    
    # Select 200 for test
    test_split = shuffled.select(range(200))
    test_datasets.append(test_split)
    
    # Keep the rest for train
    train_split = shuffled.select(range(200, len(shuffled)))
    train_datasets.append(train_split)

# Combine all splits
train_dataset = concatenate_datasets(train_datasets)
test_dataset = concatenate_datasets(test_datasets)

print("Train dataset:", train_dataset)
print("Test dataset:", test_dataset)
```

```python
train_dataset.save_to_disk("./panasonic_qa_v1_train/")
test_dataset.save_to_disk("./panasonic_qa_v1_test/")
```
